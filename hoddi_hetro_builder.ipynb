```python
import os
import torch
import pandas as pd
from torch_geometric.data import HeteroData
import pickle

BASE_DIR = "Data"
DICT_DIR = os.path.join(BASE_DIR, "dictionary")

# Load combos.pkl to get consistent drug and side effect IDs
with open(os.path.join(BASE_DIR, "combos.pkl"), "rb") as f:
    ds = pickle.load(f)

# Initialize hetero graph
data = HeteroData()

# ---- Node Features ----
num_drugs = len(ds["drug_ids"])
num_sidefx = len(ds["se_ids"])

# Give each drug/side-effect a one-hot feature vector
data["drug"].x = torch.eye(num_drugs)
data["side_effect"].x = torch.eye(num_sidefx)

# ---- Edges (drug -> side_effect) ----
# Load from HGNN/positive_samples or combos.pkl if edges not separate
edges = []
for ex in ds["examples"]:
    for se in ex["label_idxs"]:
        edges.append([ex["drug_idx"], se])

edges = torch.tensor(edges, dtype=torch.long).T  # shape [2, num_edges]
data["drug", "causes", "side_effect"].edge_index = edges

# ---- Save ----
torch.save(data, os.path.join(BASE_DIR, "hoddi_hetero.pt"))
print("✅ Saved HODDI hetero graph:", data)
print("Node types:", data.node_types)
print("Edge types:", data.edge_types)
```
